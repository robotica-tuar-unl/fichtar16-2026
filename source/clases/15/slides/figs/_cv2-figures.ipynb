```python
import numpy as np
import matplotlib.pyplot as plt

import cv2
```

```python
from glob import glob
images = np.asarray([cv2.imread(file) for file in glob("image-??.png")])
```

```python
# Determine the number of rows and columns for the subplot grid
num_images = len(images)
rows = 2 # Adjust as needed for your layout
cols = num_images // rows # Adjust as needed for your layout

plt.figure(figsize=(10, 5)) # Adjust figure size as needed

for i, img in enumerate(images):
    plt.subplot(rows, cols, i + 1) # Create subplot for each image
    plt.imshow(img)
    plt.axis('off') # Hide axes for cleaner display

plt.suptitle(f"Ejemplos - Tamaño: {images[0].shape[:2]}", fontsize=13, y=0.92)

plt.tight_layout() # Adjust subplot parameters for a tight layout
# plt.savefig("examples.svg", format="svg", bbox_inches="tight")
plt.show() # Display the plot
```

```python
selected = [0, 2, 5, 6]
```

```python
RESIZE_RATIO = 0.25

plt.figure(figsize=(10, 5)) # Adjust figure size as needed

for i, img in enumerate(images[selected]):
    # Downsamplig
    img_ds = cv2.resize(img, None, 1, RESIZE_RATIO, RESIZE_RATIO, cv2.INTER_CUBIC)
    
    plt.subplot(rows, cols, i + 1) # Create subplot for each image
    plt.imshow(img)
    plt.axis('off') # Hide axes for cleaner display
    
    plt.subplot(rows, cols, i + 1 + cols) # Create subplot for each image
    plt.imshow(img_ds)
    plt.axis('off') # Hide axes for cleaner display

plt.figtext(0.5, 0.95, f"Original {img.shape}", ha="center", fontsize=13)
plt.figtext(0.5, 0.48, f"Re-escalado {img_ds.shape}", ha="center", fontsize=13)

plt.tight_layout() # Adjust subplot parameters for a tight layout
# plt.savefig("downsampling.svg", format="svg", bbox_inches="tight")
plt.show() # Display the plot
```

```python
plt.figure(figsize=(10, 5)) # Adjust figure size as needed

for i, img in enumerate(images[selected]):
    # Downsamplig
    img_ds = cv2.resize(img, None, 1, RESIZE_RATIO, RESIZE_RATIO, cv2.INTER_CUBIC)
    
    # Convert to HSV
    imageHSV = cv2.cvtColor(img_ds, cv2.COLOR_BGR2HSV)
    
    # Mask
    lower = np.array([ 0,  0,  0])
    upper = np.array([255, 255, 30])
    mask = cv2.inRange(imageHSV, lower, upper)
    
    plt.subplot(rows, cols, i + 1) # Create subplot for each image
    plt.imshow(img)
    plt.axis('off') # Hide axes for cleaner display
    
    plt.subplot(rows, cols, i + 1 + cols) # Create subplot for each image
    plt.imshow(mask)
    plt.axis('off') # Hide axes for cleaner display

plt.figtext(0.5, 0.95, f"Original {img.shape}", ha="center", fontsize=13)
plt.figtext(0.5, 0.48, f"Máscara {mask.shape}", ha="center", fontsize=13)

plt.tight_layout() # Adjust subplot parameters for a tight layout
# plt.savefig("mask.svg", format="svg", bbox_inches="tight")
plt.show() # Display the plot
```

```python
plt.figure(figsize=(10, 5)) # Adjust figure size as needed

for i, img in enumerate(images[selected]):
    # Downsamplig
    img_ds = cv2.resize(img, None, 1, 0.25, 0.25, cv2.INTER_CUBIC)
    
    # Convert to HSV
    imageHSV = cv2.cvtColor(img_ds, cv2.COLOR_BGR2HSV)
    
    # Mask
    lower = np.array([ 0,  0,  0])
    upper = np.array([255, 255, 30])
    mask = cv2.inRange(imageHSV, lower, upper)
    
    M = cv2.moments(mask)
    if M['m00'] > 0:
        cx = int(M['m10']/M['m00'])
        cy = int(M['m01']/M['m00'])
    
        plt.subplot(rows, cols, i + 1) # Create subplot for each image
        plt.imshow(img)
        # Draw a reference
        plt.scatter(cx/RESIZE_RATIO, cy/RESIZE_RATIO, marker="o", c="C3")
        plt.axis('off') # Hide axes for cleaner display
        
        plt.subplot(rows, cols, i + 1 + cols) # Create subplot for each image
        plt.imshow(mask)
        # Draw a reference
        if i == 0:
            plt.scatter(cx, cy, marker="o", c="C3", label="Centroide")
        else:
            plt.scatter(cx, cy, marker="o", c="C3")
        plt.axis('off') # Hide axes for cleaner display

plt.figtext(0.5, 0.95, f"Original {img.shape}", ha="center", fontsize=13)
plt.figtext(0.5, 0.48, f"Máscara {mask.shape}", ha="center", fontsize=13)

plt.gcf().legend(loc="upper right", bbox_to_anchor=(0.98, 1.0))

plt.tight_layout() # Adjust subplot parameters for a tight layout
# plt.savefig("centroids.svg", format="svg", bbox_inches="tight")
plt.show() # Display the plot
```
